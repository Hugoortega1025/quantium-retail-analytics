# Quantium Retail Analytics - Task 1
## Customer Segment Analysis: Chip Purchasing Behavior

### Business Context
Category managers in grocery/CPG retail operate under constrained shelf space, promotional budget, and range depth, and cannot treat all customers equally. Without segment-level analysis, decisions about where to invest shelf space and promotions default to guesswork rather than evidence.

### Business Challenge
Which customer segments drive the most chip category revenue, and what's actually behind that revenue — more customers, higher spend per customer, or different product preferences (brand, pack size)? The answer determines where shelf space, promotions, and range decisions should be targeted.

### Approach
Clean and merge transaction and customer segment data, then analyze segment-level sales, customer counts, spend per customer, and brand/pack-size preferences to identify the highest-value segments and the specific drivers behind their purchasing behavior.

```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
sns.set(color_codes=True)
```

```python
transactions = pd.read_excel('data/raw/QVI_transaction_data.xlsx')
customers = pd.read_csv('data/raw/QVI_purchase_behaviour.csv')
print(transactions.shape, customers.shape)
```

```python
print(transactions.head())
print(customers.head())
```

```python
print(customers.info())
print(customers.isnull().sum())
print(transactions.info())
print(transactions.isnull().sum())
```

## Data Cleaning:
The date column in the transactions dataframe needs to be reformatted to its appropriate data type.

Before a thorough analysis, the check for possible outliers was assessed and dealt with accordingly depending on the context of the data point. Real outliers may represent data errors or an account test rather than an actual customer.

```python
# Excel stores dates as serial numbers (days since 1899-12-30).
# pd.to_datetime() without unit/origin treats them as nanoseconds — producing 1970 dates.
transactions['DATE'] = pd.to_datetime(transactions['DATE'], unit='D', origin='1899-12-30')

print(transactions['DATE'].min(), transactions['DATE'].max())
# Expected range: 2018-07-01 to 2019-06-30
print(transactions['DATE'].head())
```

```python
# Check for outliers in the datasets
print(transactions['PROD_QTY'].describe())
print(transactions['TOT_SALES'].describe(),'\n')
print(transactions['PROD_NAME'].unique())

# Check the maximum value in PROD_QTY and TOT_SALES too see if same person 
print(transactions[transactions['PROD_QTY'] == 200], '\n')

print(transactions[transactions['TOT_SALES'] == transactions['TOT_SALES'].max()])

```

```python
# Check the customer who bought the 200 units of chips
print(transactions[transactions['PROD_QTY'] == 200]['LYLTY_CARD_NBR'].unique())
print(transactions[transactions['TOT_SALES'] == 650]['LYLTY_CARD_NBR'].unique())
# Same customer so lets see if thats there only transaction
print(transactions[transactions['LYLTY_CARD_NBR'] == 226000],'\nThis is their transaction history')
```

### Verdict — Test Account Removed
LYLTY_CARD_NBR 226000 purchased 200 units of chips across just two transactions — an order of magnitude above any other customer in the dataset. With only two transactions in the entire purchase history and no other activity, this is a test account rather than a genuine customer. Removed from the dataset.

```python
# Distribution of PROD_QTY and TOT_SALES
fig, axes = plt.subplots(1, 2, figsize=(15, 5))
sns.histplot(transactions['PROD_QTY'], bins=20, kde=True, ax=axes[0])
axes[0].set_title('Distribution of Product Quantity')
axes[0].set_xlabel('Product Quantity')
axes[0].set_ylabel('Frequency')

sns.histplot(transactions['TOT_SALES'], bins=50, kde=True, ax=axes[1])
axes[1].set_title('Distribution of Total Sales')
axes[1].set_xlabel('Total Sales')
axes[1].set_ylabel('Frequency')
plt.show()
```

### Non-Chip products removed
The chip category contained some dip/salsa products and was removed as the focus of the project involves chip purchases only.

```python
# Remove the outlier from the transactions dataset
transactions = transactions[transactions['LYLTY_CARD_NBR'] != 226000]
transactions['PROD_NAME'].unique()
print(transactions[transactions['PROD_NAME'].str.contains('salsa', case=False)],'\n')

# Remove the salsa transactions as they are not relevant to the analysis of chips
# Remove non-chip products
mask = (transactions['PROD_NAME'].str.contains('Dip', case=False)) | \
       (transactions['PROD_NAME'].str.contains('Salsa', case=False) & 
        transactions['PROD_NAME'].str.contains('300g', case=False))

print('Rows to remove:', mask.sum())
print(transactions[mask]['PROD_NAME'].unique())

# Remove the identified rows
transactions = transactions[~mask]
```

## Feature Engineering

### Pack Size
Extracting pack size in grams from the product name to enable analysis of whether certain customer segments prefer larger  or smaller bags.

```python
# FEATURE ENGINEERING ANALYSIS (Pack Size)
transactions['PACK_SIZE'] = transactions['PROD_NAME'].str.extract(r'(\d+)[gG]')
transactions['PACK_SIZE'] = transactions['PACK_SIZE'].astype(int)
print(transactions['PACK_SIZE'].describe())
print(np.sort(transactions['PACK_SIZE'].unique()))



```

```python
# What pack size is most popular among customers?
pack_size = transactions.groupby('PACK_SIZE')['TOT_SALES'].sum().reset_index()
pack_size = pack_size.sort_values(by='TOT_SALES', ascending=False)
plt.figure(figsize=(10, 6))
sns.barplot(x='PACK_SIZE', y='TOT_SALES', data=pack_size)
plt.title('Total Sales by Pack Size')
plt.xlabel('Pack Size (g)')
plt.ylabel('Total Sales')
plt.show()
```

### Pack Size Finding
The 175g pack is the dominant size by total revenue, consistent with it being the most common standard bag size across brands. The 270g and 330g packs over-index relative to their SKU count, indicating that large-format bags drive disproportionate revenue when stocked. Shelf space allocation should prioritize the 175g standard and large-format packs (270g+) over niche sizes below 150g, which generate minimal revenue despite occupying shelf real estate.

### Brand Name
Extracting the brand from the product name and standardizing inconsistent brand spellings to ensure accurate analysis.

```python
# Extract brand name
transactions['BRAND'] = transactions['PROD_NAME'].str.split().str[0]
print(transactions['BRAND'].unique())

#Fix the brand names to be more consistent
transactions['BRAND'] = transactions['BRAND'].replace({
    'Dorito': 'Doritos',
    'Natural' : 'Natural Chip Co',
    'Smith': 'Smiths',
    'RRD': 'Red Rock Deli',
    'Red' : 'Red Rock Deli',
    'Snbts': 'Sunbites',
    'NCC': 'Natural Chip Co',
    'Infzns': 'Infuzions',
    'GrnWves':'Grain Waves',
    'Grain':'Grain Waves',
    'WW' : 'Woolworths',
})

print('Updated brand names:\n', transactions['BRAND'].unique())
```

## Merging Datasets
Combining transaction data with customer segment data on loyalty card number to enable segment-level analysis.

```python
# merge trnsactions and customers datasets on LYLTY_CARD_NBR
merged_data = pd.merge(transactions, customers, on='LYLTY_CARD_NBR', how='inner')
print(merged_data.head())

print('\n')
print(merged_data.isnull().sum())
print('Unique customers:', merged_data['LYLTY_CARD_NBR'].nunique())
print('Unique lifestages:', merged_data['LIFESTAGE'].unique())
print('Unique premium segments:', merged_data['PREMIUM_CUSTOMER'].unique())
```

## Customer Segment Analysis



### Total Sales by Segment
Each bar is one customer segment (a Lifestage like "Older Families" combined with a spend tier: Budget, Mainstream, or Premium). This chart only answers "who brings in the most total revenue" — it can't tell you if that's from lots of customers or big individual spenders.


```python
# Analysis of which segments spend the most on chips

segment_sales = merged_data.groupby(['LIFESTAGE', 'PREMIUM_CUSTOMER'])['TOT_SALES'].sum().reset_index()
segment_sales = segment_sales.sort_values(by='TOT_SALES', ascending=False)
print(segment_sales)

lifestage_order = segment_sales.groupby('LIFESTAGE')['TOT_SALES'].sum().sort_values(ascending=False).index
plt.figure(figsize=(12, 6))
sns.barplot(x='LIFESTAGE', y='TOT_SALES', hue='PREMIUM_CUSTOMER', data=segment_sales,
            order=lifestage_order, hue_order=['Budget', 'Mainstream', 'Premium'],
            palette=['steelblue', 'green', 'gold'])
plt.title('Total Sales by Lifestage and Premium Customer Segment')
plt.xlabel('Lifestage')
plt.ylabel('Total Sales')
plt.xticks(rotation=45, fontsize=10)
plt.tight_layout()
plt.show()
plt.show()
```

### Sales per Customer
Total sales alone can be misleading because a segment with more customers will naturally spend more overall. A normalizing approach was employed to reveal which segments spend the most per person.


```python
# Normalise the total sales by the number of customers in each segment to get average sales per customer

segment_customers = merged_data.groupby(['LIFESTAGE', 'PREMIUM_CUSTOMER'])['LYLTY_CARD_NBR'].nunique().reset_index()
segment_customers.columns = ['LIFESTAGE', 'PREMIUM_CUSTOMER', 'NUM_CUSTOMERS']

segment_analysis = pd.merge(segment_sales, segment_customers, on=['LIFESTAGE', 'PREMIUM_CUSTOMER'])
segment_analysis['AVG_SALES_PER_CUSTOMER'] = segment_analysis['TOT_SALES'] / segment_analysis['NUM_CUSTOMERS']
segment_analysis = segment_analysis.sort_values(by='AVG_SALES_PER_CUSTOMER', ascending=False)
print(segment_analysis.head(10))

lifestage_order = segment_analysis.groupby('LIFESTAGE')['AVG_SALES_PER_CUSTOMER'].sum().sort_values(ascending=False).index
plt.figure(figsize=(12, 6))
sns.barplot(x='LIFESTAGE', y='AVG_SALES_PER_CUSTOMER', hue='PREMIUM_CUSTOMER', data=segment_analysis,
            order=lifestage_order, hue_order=['Budget', 'Mainstream', 'Premium'],
            palette=['steelblue', 'green', 'gold'])
plt.title('Average Sales per Customer by Lifestage and Premium Customer Segment')
plt.xlabel('Lifestage')
plt.ylabel('Average Sales per Customer')
plt.xticks(rotation=45, fontsize=10)
plt.tight_layout()
plt.show()
```

### Average Bags per Customer
Looking at how many bags of chips each segment buys per year to understand volume-driven vs price-driven sales differences.

```python
# Average number of chips (bags) bought per customer by segment

bags_per_customer = merged_data.groupby(['LIFESTAGE', 'PREMIUM_CUSTOMER']).agg(
    TOTAL_BAGS = ('PROD_QTY', 'sum'),
    NUM_CUSTOMERS = ('LYLTY_CARD_NBR','nunique')
).reset_index()

bags_per_customer['AVG_BAGS_PER_CUSTOMER'] = bags_per_customer['TOTAL_BAGS'] / bags_per_customer['NUM_CUSTOMERS']
bags_per_customer = bags_per_customer.sort_values('AVG_BAGS_PER_CUSTOMER', ascending=False)

lifestage_order = bags_per_customer.groupby('LIFESTAGE')['AVG_BAGS_PER_CUSTOMER'].sum().sort_values(ascending=False).index
plt.figure(figsize=(12, 6))
sns.barplot(x='LIFESTAGE', y='AVG_BAGS_PER_CUSTOMER', hue='PREMIUM_CUSTOMER', data=bags_per_customer,
            order=lifestage_order, hue_order=['Budget', 'Mainstream', 'Premium'],
            palette=['steelblue', 'green', 'gold'])
plt.title('Average Bags (Chips) per Customer by Segment')
plt.xlabel('Lifestage')
plt.ylabel('Avg Bags per Customer')
plt.xticks(rotation=45, fontsize=10)
plt.tight_layout()
plt.show()
```

### Reading the Three Charts Together
Looking at all three side by side: **Budget Older Families** leads on both total revenue and average bags per customer — a large, loyal group that buys a lot and buys often, so their leadership on the first chart is genuine, not just a headcount effect.

**Mainstream Young Singles/Couples** tell a different story: high on the total-sales chart, but much lower on spend-per-customer and bags-per-customer. That combination means this segment's revenue is coming from having a lot of customers who each buy a little, not from individually big spenders. That's the pattern worth investigating further — it raises the question of what they're buying, since it's clearly not "more of the same" behavior as the top segment. That's exactly what the brand and pack size charts below dig into.

```python
# Brand preference - Mainstream Young Singles/Couples vs rest of population
target = merged_data[
    (merged_data['LIFESTAGE'] == 'YOUNG SINGLES/COUPLES') & 
    (merged_data['PREMIUM_CUSTOMER'] == 'Mainstream')
]
other = merged_data[~(
    (merged_data['LIFESTAGE'] == 'YOUNG SINGLES/COUPLES') & 
    (merged_data['PREMIUM_CUSTOMER'] == 'Mainstream')
)]

# Brand share within each group
target_brand = target.groupby('BRAND').size() / len(target)
other_brand = other.groupby('BRAND').size() / len(other)

# How much more likely this segment buys a brand vs everyone else
brand_preference = (target_brand / other_brand).sort_values(ascending=False)
print('Brand affinity for Mainstream Young Singles/Couples:')
print(brand_preference.head(10))

plt.figure(figsize=(12, 6))
brand_preference.head(10).plot(kind='bar', color = 'salmon')
plt.title('Brand Preference — Mainstream Young Singles/Couples vs Population')
plt.xlabel('Brand')
plt.ylabel('Preference Score (>1 = buys more than average')
plt.xticks(rotation=45)
plt.axhline(y=1, color='black', linestyle='--', label='Population average')
plt.legend()
plt.tight_layout()
plt.savefig('images/brand_preference.png', bbox_inches='tight', dpi=150)
plt.show()
```

```python
# Pack size preference 
target_pack = target.groupby('PACK_SIZE').size() / len(target)
other_pack = other.groupby('PACK_SIZE').size() / len(other)

pack_preference = (target_pack / other_pack).sort_values(ascending=False)
print('Pack size preference for Mainstream Young Singles/Couples:')
print(pack_preference)

plt.figure(figsize=(12, 6))
pack_preference.plot(kind='bar', color = 'salmon')
plt.title('Pack Size Preference — Mainstream Young Singles/Couples vs Population')
plt.xlabel('Pack Size (g)')
plt.ylabel('Preference Score (>1 = buys more than average')
plt.xticks(rotation=0)
plt.axhline(y=1, color='red', linestyle='--', label='Population average')
plt.legend()
plt.tight_layout()
plt.show()
```

### Pack Size Preference — Mainstream Young Singles/Couples

The clearest pattern is a preference for large-format packs: 270g (1.27x), 380g (1.26x), and 330g (1.23x) are this segment's top three pack sizes by a clear margin. There's also a secondary lift across a mid-range cluster — 134g, 210g, 110g, 135g, and 250g all score between 1.13x and 1.18x. Notably, this is *not* a "small vs. large" story: the smallest packs (90g at 0.52x, 70g at 0.51x) are among the segment's most under-indexed choices, not preferred ones. Promotional and shelf strategy for this segment should lean into large-format packs first, with secondary support for the 110–135g and 210–250g range — not the smallest sizes.

### Segment Summary — Key Findings

**Budget Older Families** are the highest-revenue segment in absolute terms, driven by high purchase frequency and large household volume. They represent the most commercially significant group for total chip category revenue.

**Mainstream Young Singles/Couples** rank second in total sales but lowest in average spend per customer — meaning the segment's revenue comes from volume of customers rather than individual spend. The brand and pack size analysis reveals a distinct purchasing profile: elevated affinity for a mix of craft and mainstream snack brands (Tyrrells, Twisties, and Doritos lead the list) rather than a clean boutique-vs-mass-market split, and a preference for large-format packs (270g+) with secondary interest in the 110–135g and 210–250g range — not the smallest packs, which this segment actively under-indexes on.

**Business implication:** These two segments require different strategies. Budget Older Families should be targeted with multi-buy promotions on high-volume standard SKUs. Mainstream Young Singles/Couples respond to brand differentiation and pack-size variety — shelf placement and promotional priority should reflect the specific brands and sizes this segment over-indexes on, particularly large-format packs and the flavor-forward brands (both craft and mainstream) that led the affinity list.
